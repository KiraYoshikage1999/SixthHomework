```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

X = np.array([2, 4, 6, 8, 10, 12])
Y = np.array([3.0, 5.5, 7.0, 10.0, 12.0, 14.5])

data = pd.DataFrame({"screens": X, "cost_thousands": Y})
display(data)

w = 0.0
b = 0.0
learning_rate = 0.001
epochs = 5000
n = len(X)

error_history = []

for epoch in range(epochs):
    y_pred = w * X + b
    error = y_pred - Y

    dw = (2 / n) * np.sum(error * X)
    db = (2 / n) * np.sum(error)

    w = w - learning_rate * dw
    b = b - learning_rate * db

    error_history.append(np.mean(error ** 2))

print(f"Price per screen (w): ${w:.2f}k")
print(f"Initial cost (b): ${b:.2f}k")

new_screens = 9
predicted_cost = w * new_screens + b
print(f"Prediction for {new_screens} screens: ${predicted_cost:.2f}k")

plt.scatter(X, Y, color="blue", label="Previous projects")

plt.plot(X, w * X + b, color="red", label="Linear model")
plt.scatter(new_screens, predicted_cost, color="green", label="New project")

plt.xlabel("Number of screens")

plt.ylabel("Cost ($k)")
plt.title("Mobile app cost estimation")

plt.legend()
plt.show()

plt.figure(figsize=(10, 5))
plt.plot(range(1, epochs + 1), error_history, color="crimson", label="Error")

plt.yscale("log")
plt.xlabel("Training step")

plt.ylabel("Mean Squared Error (log scale)")
plt.title("Error reduction during training")

plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.legend()

plt.tight_layout()

plt.show()
```
